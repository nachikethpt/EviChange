# EviChange — GEOBench-VLM temporal tasks (outside yardstick for the Geo-VLM choice)

(Danish et al., ICCV 2025): crop type, disaster type, farm-pond change, land use, damaged-building count.

runtime (not to your computer). The dataset card names no licence (the code is Apache-2.0): use it for
evaluation and cite the paper.

**Runtime:** L4 is enough. Run Step 1, then Step 2 for one model; for the other model use
`Runtime → Restart session`, then Step 1 again and Step 2 with the other model name.
